```python

from sklearn.neighbors import KNeighborsClassifier

# Training data
X = [[1], [2], [3], [4], [5], [6]]
y = [0, 0, 0, 1, 1, 1]

# Create K-NN model
model = KNeighborsClassifier(n_neighbors=3)

# Train the model
model.fit(X, y)

# Test the model
result = model.predict([[2.5]])

print("Predicted Class:", result[0])
```
