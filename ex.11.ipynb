```python
from sklearn.tree import DecisionTreeClassifier

# Training data
X = [
    [750, 50000, 2],
    [680, 30000, 1],
    [600, 20000, 3],
    [800, 80000, 1],
    [550, 15000, 4],
    [720, 60000, 2]
]

# Credit score categories
y = ['Good', 'Average', 'Poor', 'Good', 'Poor', 'Good']

# Create and train the model
model = DecisionTreeClassifier(random_state=0)
model.fit(X, y)

# Predict a new customer's credit score
new_customer = [[750, 50000, 2]]

prediction = model.predict(new_customer)

print("Credit Score Category:", prediction[0])
```
