```python
from sklearn.linear_model import LinearRegression

# Training data
X = [[1], [2], [3], [4], [5]]
y = [900000, 800000, 700000, 600000, 500000]

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict car price
age = [[3]]
predicted_price = model.predict(age)

print("Car Age:", age[0][0], "years")
print("Predicted Car Price: Rs.", predicted_price[0])
```
