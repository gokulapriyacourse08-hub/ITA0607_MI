```python

size = [1000, 1500, 2000, 2500, 3000]
price = [20, 30, 40, 50, 60]

n = len(size)

mx = sum(size) / n
my = sum(price) / n

slope = sum((size[i]-mx)*(price[i]-my) for i in range(n)) / sum((x-mx)**2 for x in size)

intercept = my - slope * mx

new_size = 2200
predicted_price = slope * new_size + intercept

print("House Size:", new_size, "sq.ft")
print("Predicted House Price:", predicted_price, "Lakhs")
```
