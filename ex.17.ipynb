```python

from sklearn.neighbors import KNeighborsClassifier

X = [
    [2, 32, 3000],
    [3, 64, 3500],
    [4, 64, 4000],
    [4, 128, 4500],
    [6, 128, 5000],
    [8, 256, 6000]
]

y = [0, 0, 1, 1, 2, 3]

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

ram = 4
memory = 128
battery = 4500

result = model.predict([[ram, memory, battery]])

print("RAM:", ram, "GB")
print("Memory:", memory, "GB")
print("Battery:", battery, "mAh")
print("Predicted Price Range:", result[0])
```
