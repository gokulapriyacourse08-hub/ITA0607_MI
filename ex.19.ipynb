```python

from sklearn.naive_bayes import GaussianNB

X = [
    [25, 30000],
    [30, 40000],
    [35, 50000],
    [40, 60000],
    [45, 70000],
    [22, 20000],
    [50, 80000],
    [28, 35000]
]

y = [0, 0, 1, 1, 1, 0, 1, 0]

model = GaussianNB()
model.fit(X, y)

age = 38
salary = 55000

result = model.predict([[age, salary]])

print("Age:", age)
print("Salary:", salary)

if result[0] == 1:
    print("Loan Status: Approved")
else:
    print("Loan Status: Not Approved")
```
