```python

month = [1, 2, 3, 4, 5]
sales = [100, 200, 300, 400, 500]

n = len(month)

mx = sum(month) / n
my = sum(sales) / n

slope = sum((month[i]-mx)*(sales[i]-my) for i in range(n)) / sum((x-mx)**2 for x in month)

intercept = my - slope * mx

future_month = 6
prediction = slope * future_month + intercept

print("Future Month:", future_month)
print("Predicted Sales:", prediction)
```
